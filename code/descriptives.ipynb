```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from paths import data_path
```

```python
# are sales higher around oil price shocks, paycheck days and the 2016 earthquake?
df=pd.read_csv(f'{data_path}/train.csv')
df['date']=pd.to_datetime(df['date'])
df['day']=df['date'].dt.day
```

```python
# check sales around paycheck days
df_day=df.groupby('day')['sales'].mean().reset_index()
sns.lineplot(x='day', y='sales', data=df_day)
plt.xticks(df_day['day'], rotation=45)
for d in [15, 30, 31]:
	plt.axvline(d, color='red')

```

```python
# check sales around april 16, 2016 earthquake and holidays

df_date = df.groupby('date')['sales'].mean().reset_index()
sns.lineplot(x='date', y='sales', data=df_date)
plt.axvline(pd.to_datetime('2016-04-16'), color='red', linestyle='--', label='Earthquake')
plt.xticks(rotation=45)

```

```python
# plot oil prices
df=pd.read_csv(f'{data_path}/oil.csv')
df['date']=pd.to_datetime(df['date'])
df_date = df.groupby('date')['dcoilwtico'].mean().reset_index()
sns.lineplot(x='date', y='dcoilwtico', data=df_date)
plt.xticks(rotation=45)
```
